```python
import pandas as pd

df = pd.read_csv(
    "data/대구광역시 남구_불법주정차단속현황_20251031.csv",
    encoding="cp949"
)

print("데이터 크기:", df.shape)

print("\n컬럼명")
print(df.columns.tolist())

print("\n앞 5개")
print(df.head())

print("\n데이터 타입")
print(df.info())

print("\n결측치")
print(df.isnull().sum())
```

```python
df["위반일자"] = pd.to_datetime(df["위반일자"])

print(df.info())
```

```python
# 연도
df["연도"] = df["위반일자"].dt.year

# 월
df["월"] = df["위반일자"].dt.month

# 요일
df["요일"] = df["위반일자"].dt.day_name()

# 시간
df["시간"] = df["위반시간"].str.split(":").str[0].astype(int)

df.head()
```
