```python
import pandas as pd
import glob
import os

files = glob.glob("data/*.csv")

for file in files:
    print("=" * 80)
    print("파일명:", os.path.basename(file))

    try:
        df = pd.read_csv(file, encoding="cp949")

        print("행, 열:", df.shape)
        print("컬럼:", df.columns.tolist())
        print()
        print(df.head(3))

    except Exception as e:
        print("읽기 오류:", e)
```
