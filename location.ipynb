```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df_2020 = pd.read_csv(
    "data/parking_2020.csv",
    encoding="utf-8"
)

df_hotspot = pd.read_csv(
    "data/parking_hotspot_2021.csv",
    encoding="utf-8"
)

df_2024 = pd.read_csv(
    "data/parking_violation_2024.csv",
    encoding="cp949"
)

df_2026 = pd.read_csv(
    "data/parking_violation_2026.csv",
    encoding="utf-8"
)
```

```python
# 장소 

top10_places_2024 = df_2024['위반장소'].value_counts().head(10)

top10_places_2024
```

```python

plt.figure(figsize=(12, 6))
sns.barplot(x=top10_places_2024.values, y=top10_places_2024.index, palette='viridis')
plt.title('2024년 불법 주정차 단속 최다 적발 장소', fontsize=16, pad=15)
plt.xlabel('단속 건수', fontsize=12)
plt.ylabel('위반 장소', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
